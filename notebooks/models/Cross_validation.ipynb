Cross-validation sur le meilleur modèle (XGBoost)

```python
import pandas as pd
import numpy as np
from xgboost import XGBClassifier
from sklearn.model_selection import cross_val_score
df = pd.read_csv("../../data/weatherAUS_ForModel.csv")

X = df.drop(columns=["RainTomorrow", "Date", "SeasonName"], errors="ignore")
y = df["RainTomorrow"]
```

```python

neg = np.sum(y == 0)
pos = np.sum(y == 1)
ratio = neg / pos

model = XGBClassifier(
    n_estimators=300,
    learning_rate=0.1,
    max_depth=5,
    scale_pos_weight=ratio,
    random_state=42,
    eval_metric="logloss"
)

scores = cross_val_score(model, X, y, cv=5, scoring="roc_auc")

print("Scores :", scores)
print("Moyenne :", scores.mean())
print("Écart-type :", scores.std())
```

Avant validation croisée, le modèle obtenait un ROC-AUC d’environ 0.89, ce qui pouvait refléter une estimation optimiste des performances.
Après application d’une validation croisée à 5 plis, le ROC-AUC moyen est de 0.83 avec un écart-type de 0.037, indiquant une performance globalement satisfaisante mais une variabilité modérée selon les partitions des données.
Cette analyse montre que la validation croisée permet d’obtenir une estimation plus fiable et réaliste des performances du modèle.
